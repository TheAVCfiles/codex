# MythOS Local Quickstart\n\nThis notebook talks to the local FastAPI service (`server.py`) and demonstrates the portable `/generate` and `/align` endpoints.

```python
import requests\nBASE_URL = "http://127.0.0.1:8000"
```

```python
generate_payload = {"prompt": "Summarize local-first alignment goals", "max_tokens": 64}\nrequests.post(f"{BASE_URL}/generate", json=generate_payload, timeout=20).json()
```

```python
align_payload = {\n    "integrity": 0.9,\n    "coherence": 0.88,\n    "drift": 0.15,\n    "incidents": 3,\n    "recoveries": 2\n}\nrequests.post(f"{BASE_URL}/align", json=align_payload, timeout=20).json()
```
